```python
import pandas as pd
data={
    'student_name': ['sri','hari','mahesh','karthi'],
    'dept': ['cs','tamil','maths','science'],
    'score': [60,70,80,90]

}
df=pd.DataFrame(data)
```

```python
print(df['score'].sum())
print(df['score'].mean())
print(df['score'].min())
print(df['score'].max())
print(df['score'].count())
print(df['score'].median())
```

```python
df['score'].agg(['sum','mean','min','max','count'])
```

```python
print(df['score'].var())
```

```python
print(df['score'].std())
```
